# Implement PCA using scikit-learn

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# --------------------------------------------
# Step 1: Original Dataset
# --------------------------------------------

employees = ["A", "B", "C", "D", "E"]

X = np.array([
    [2, 65, 70],
    [4, 80, 78],
    [5, 85, 82],
    [6, 90, 88],
    [8, 98, 95]
])

print("Original Dataset")
print(X)
```

```python
# --------------------------------------------
# Step 2: Standardised the data
# --

scaler = StandardScaler()
X_std = scaler.fit_transform(X)

print("\n Standardized dataset")
print(np.round(X_std, 4))
```

Minor differences may occur because Python retains full floating-point precisions, whereas, the manual calculations were rounded to four decimal places

```python
# --------------------------------------------
# Step 3: Perform the PCA
# --

pca = PCA(n_components = 3) # number of target dimensions or the principal components to keep after reducing the dataset's size
X_pca = pca.fit_transform(X_std)


# --------------------------------------------
# Step 4: Display PCA Results
# --------------------------------------------

print("\nEigenvalues")
print(np.round(pca.explained_variance_, 6))

print("\nExplained Variance Ratio")
print(np.round(pca.explained_variance_ratio_, 6))

print("\nPrincipal Components (Eigenvectors)")
print(np.round(pca.components_, 6))

print("\nReduced Dataset")
print(np.round(X_pca, 4))
```

Comparison: The eigenvalues obtained using scikit-learn differ from the manually computed values because the manual solution used a normalised covariance (correlation) matrix with unit variances, whereas scikit-learn computes PCA using the sample covariance matrix of the standardised data. Consequently, the Python eigenvalues are scaled by a constant factor. Despite this difference, both approaches identify the same dominant principal component, and the explained variance ratios and overall conclusions remain consistent
